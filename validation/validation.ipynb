```python
import pyvista as pv
import pandas as pd
import matplotlib.pyplot as plt
reference = pv.read("../resources/final_output_30s.vts")
exp = pv.read("../results/Explicit_Euler/output_3000.vti")
rk4= pv.read("../results/Runge-Kutta/output_3000.vti")
imp=pv.read("../results/varyDt/dt_0.01/output_3000.vti")
```

```python
# Extracting C1 and C2 from the reference solution
ref_C1=reference["C1"]
ref_C2=reference["C2"]
# Extracting C1 and C2 from the explicit_euler solution
exp_C1=exp["C1"]
exp_C2=exp["C2"]
# Extracting C1 and C2 from the implicit_euler solution
imp_C1=imp["C1"]
imp_C2=imp["C2"]
# Extracting C1 and C2 from the rk4 solution
rk4_C1=rk4["C1"]
rk4_C2=rk4["C2"]
```

```python
#Defining the error functions here
import numpy as np

# l2 norm
def L2(a,b):
    return np.sqrt(np.mean((a-b)**2))
# Maximum error
def linf(a,b):
    return np.max(np.abs(a-b))
# Mean Absolute Error
def MAE(a,b):
    return np.mean(np.abs(a,b))
```

```python
comparison=pd.DataFrame({
    "Method": ["Explicit Euler", "RK4", "Implicit Euler"], 
    "L2(C1)":[
        L2(exp_C1,ref_C1),
        L2(rk4_C1,ref_C1),
        L2(imp_C1,ref_C1)
    ],

    "L∞(C2)": [
        linf(exp_C1,ref_C1),
        linf(rk4_C1,ref_C1),
        linf(imp_C1,ref_C1) 
    ],

    "MAE(C1)": [
        MAE(exp_C1,ref_C1),
        MAE(rk4_C1,ref_C1),
        MAE(imp_C1,ref_C1)
    ],

    "L2(C2)": [
        L2(exp_C2, ref_C2),
        L2(rk4_C2, ref_C2),
        L2(imp_C2, ref_C2)  
    ],

    "L∞(C2)": [
        linf(exp_C2, ref_C2),
        linf(rk4_C2, ref_C2),
        linf(imp_C2, ref_C2)
    ],

    "MAE(C2)": [
        MAE(exp_C2, ref_C2),
        MAE(rk4_C2, ref_C2),
        MAE(imp_C2, ref_C2)
    ]
})
```

```python
comparison.style.format("{:.8e}", subset=comparison.columns[1:])
```

```python
# Reshaping the arrays

nx = 40
ny = 40

ref_C1 = ref_C1.reshape((ny, nx))
ref_C2 = ref_C2.reshape((ny, nx))

exp_C1 = exp_C1.reshape((ny, nx))
exp_C2 = exp_C2.reshape((ny, nx))

rk4_C1 = rk4_C1.reshape((ny, nx))
rk4_C2 = rk4_C2.reshape((ny, nx))

imp_C1 = imp_C1.reshape((ny, nx))
imp_C2 = imp_C2.reshape((ny, nx))
```

```python
# Adding the color bars by computing the commom limits
vmin1 = min(ref_C1.min(), exp_C1.min(), rk4_C1.min(), imp_C1.min())
vmax1 = max(ref_C1.max(), exp_C1.max(), rk4_C1.max(), imp_C1.max())

vmin2 = min(ref_C2.min(), exp_C2.min(), rk4_C2.min(), imp_C2.min())
vmax2 = max(ref_C2.max(), exp_C2.max(), rk4_C2.max(), imp_C2.max())
```

```python
# Create the figure
fig, axs = plt.subplots(2, 4, figsize=(18, 9))
fig.suptitle("Concentration fields at t = 30 s for all methods",
             fontsize=18)

titles = ["Reference", "ExpEuler", "RK4", "ImpEuler"]

C1_list = [ref_C1, exp_C1, rk4_C1, imp_C1]
C2_list = [ref_C2, exp_C2, rk4_C2, imp_C2]

for i in range(4):

    im1=axs[0,i].contourf(C1_list[i], levels=10, cmap="turbo",vmin=vmin1,vmax=vmax1)
    axs[0,i].set_title(titles[i] + " C1")
    axs[0,i].set_xlabel("i (x)")
    axs[0,i].set_ylabel("j (y)")
    

    im2=axs[1,i].contourf(C2_list[i], levels=10, cmap="viridis",vmin=vmin2,vmax=vmax2)
    axs[1,i].set_title(titles[i] + " C2")
    axs[1,i].set_xlabel("i (x)")
    axs[1,i].set_ylabel("j (y)")

# Leave some room on the right
plt.subplots_adjust(right=0.88)
# Create separate axes for the colorbars
cax1 = fig.add_axes([0.90, 0.57, 0.015, 0.30])  # [left, bottom, width, height]
cax2 = fig.add_axes([0.90, 0.13, 0.015, 0.30])

#Draw the colorbars
cbar1 = fig.colorbar(im1,cax=cax1)
cbar1.set_label("C1")

cbar2 = fig.colorbar(im2,cax=cax2)
cbar2.set_label("C2")

plt.subplots_adjust(
    left=0.06,
    right=0.88,
    bottom=0.07,
    top=0.92,
    wspace=0.18,
    hspace=0.22
)

#Save the figure
plt.savefig("comparison of all methods.png", dpi=300,bbox_inches="tight")

plt.show()
```
